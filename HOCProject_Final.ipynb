###Lien Holder

```python
import matplotlib.patches as mpatches

# --- Define Lien Holder categories and variations ---
lien_categories = [
    'City of Camden',
    'Pro Cap 8 FBO FIRSTRUST BANK',
    'Open Taxes',
    'The Phoenix Development Group',
    'Other'
]

# Define colors for categories
lien_colors = [
    '#1f78b4',  # Blue
    '#ff7f00',  # Orange
    '#e31a1c',  # Red
    '#33a02c',  # Green
    '#6a3d9a'   # Purple
]

# Create mapping dictionary for all known variations
lien_map = {
    # City of Camden variations
    'City of Camden': 'City of Camden',
    'City of camden': 'City of Camden',
    'City of Camden ': 'City of Camden',

    # Pro Cap variations
    'Pro Cap 8 FBO FIRSTRUST BANK': 'Pro Cap 8 FBO FIRSTRUST BANK',
    'Pro Cap 8 FBO FIRSTRUST BANK ': 'Pro Cap 8 FBO FIRSTRUST BANK',

    # Open Taxes variations
    'open taxes': 'Open Taxes',
    'open taxes ': 'Open Taxes',
    'Open taxes': 'Open Taxes',
    'Open Taxes': 'Open Taxes',

    # Phoenix Development Group variations
    'The Phoenix Development Group': 'The Phoenix Development Group',
    'The Phoenix Development Group ': 'The Phoenix Development Group'
}

# Function to map lien holders to canonical category, "Other" if not mapped and not empty
def map_lien_category(holder):
    if pd.isna(holder) or holder.strip() == '':
        return None  # Keep NaN for empty values
    return lien_map.get(holder, 'Other')

# Apply mapping
wf_south['Lien_Category'] = wf_south['LienHolder'].map(map_lien_category)
lansdowne_area['Lien_Category'] = lansdowne_area['LienHolder'].map(map_lien_category)

# --- Create color mapping ---
lien_cmap = dict(zip(lien_categories, lien_colors))

def map_lien_color(category):
    return lien_cmap.get(category, '#d3d3d3')  # light gray for unmapped / empty

# Assign colors
wf_south['color'] = wf_south['Lien_Category'].map(map_lien_color)
lansdowne_area['color'] = lansdowne_area['Lien_Category'].map(map_lien_color)

# --- Main plot ---
fig = plt.figure(figsize=(6, 7))
main_ax = fig.add_axes([0.1, 0.1, 0.8, 0.75])

wf_south.plot(ax=main_ax, color=wf_south['color'], edgecolor='black', linewidth=0.5, alpha=0.8)
ctx.add_basemap(main_ax, source=ctx.providers.OpenStreetMap.Mapnik)
main_ax.set_xlim(*wf_south.total_bounds[[0, 2]])
main_ax.set_ylim(*wf_south.total_bounds[[1, 3]])
main_ax.set_title("Waterfront South – Lien Holder Categories", fontsize=14, fontweight='bold')

# --- Inset plot (optional) ---
inset_ax = fig.add_axes([0.5, 0.9, 0.4, 0.12])
lansdowne_area.plot(ax=inset_ax, color=lansdowne_area['color'], edgecolor='black', linewidth=0.5, alpha=0.8)
ctx.add_basemap(inset_ax, source=ctx.providers.OpenStreetMap.Mapnik)
inset_ax.set_xlim(*lansdowne_area.total_bounds[[0, 2]])
inset_ax.set_ylim(*lansdowne_area.total_bounds[[1, 3]])
inset_ax.set_xticks([])
inset_ax.set_yticks([])
inset_ax.set_title("Area Around Lansdowne Ave", fontsize=10, fontweight='bold')

# --- Legend with counts ---
lien_counts = pd.concat([wf_south['Lien_Category'], lansdowne_area['Lien_Category']]).value_counts().to_dict()

legend_patches = [
    mpatches.Patch(color=color, label=f"{category} ({lien_counts.get(category, 0)})")
    for category, color in lien_cmap.items()
]

legend_ax = fig.add_axes([0.09, 0.92, 0.3, 0.1])
legend_ax.axis('off')
legend = legend_ax.legend(
    handles=legend_patches,
    loc='center left',
    frameon=False,
    title='Lien Holder Category',
    fontsize=8,
    title_fontsize='small'
)
legend.get_title().set_fontweight('bold')

main_ax.set_xticks([])
main_ax.set_yticks([])

plt.show()
```
